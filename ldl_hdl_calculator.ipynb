```python
#creating an LDL and HDL calculator
def calculate_ldl(total_cholesterol,hdl,triglyceride):
  ldl= total_cholesterol-hdl-(triglyceride/2.2)
  return ldl

def calculate_hdl(total_cholesterol, ldl, triglyceride):
  hdl= total_cholesterol-ldl- (triglyceride/2.2)
  return hdl

print("CHOLESTEROL CALCULATOR")
print("1. Calculate LDL")
print("2. Calculate HDL")
choice= input("Enter chose 1 or 2: ")
if choice=="1":
  total_cholesterol= float(input("Enter total cholesterol(mmol/l): "))
  hdl= float(input("Enterl HDL(mmol/l): "))
  triglyceride= float(input("Enter triglyceride(mmol/l):"))
  result= calculate_ldl(total_cholesterol,hdl, triglyceride)
  print("calculated LDL:", round(result, 2), "mmol/l")
elif choice=="2":
  total_cholesterol= float(input("Enter total cholesterol(mmol/l): "))
  hdl= float(input("Enter HDL(mmol/l): "))
  triglyceride= float(input("Enter triglyceride(mmol/l):"))
  result= calculate_hdl(total_cholesterol, ldl, triglyceride)
  print("Calculated HDL:", round (result,2),"mmol/l")
else:
  print("Invalid Choice. Please chose 1 or 2")
```
